**Ejercicio 3.1 · FizzBuzz**

El orden importa: hay que comprobar primero el caso "múltiplo de ambos" (15), o nunca se llegaría a él.

```python
for n in range(1, 21):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```

**Ejercicio 3.2**

```python
notas = {"Ana": 8.5, "Luis": 4.0, "Marta": 6.2, "Pedro": 3.5, "Eva": 9.1}

for nombre, nota in notas.items():
    estado = "aprobado" if nota >= 5 else "suspenso"
    print(f"{nombre}: {nota} → {estado}")

aprobados = [nombre for nombre, nota in notas.items() if nota >= 5]
aprobados
```

**Ejercicio 3.3**

```python
bacterias = 10
horas = 0
while bacterias <= 1_000_000:
    bacterias *= 2
    horas += 1

print(f"Tras {horas} horas hay {bacterias:,} bacterias")
```
