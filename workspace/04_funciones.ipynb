## 4.1 Definir una función

Una función agrupa código con un nombre para poder reutilizarlo. Se define con `def`, recibe **parámetros** y devuelve un resultado con `return`.

```python
def area_rectangulo(base, altura):
    """Devuelve el área de un rectángulo."""
    return base * altura


print(area_rectangulo(4, 3))
print(area_rectangulo(10, 2.5))
```

El texto entre triples comillas es la **docstring**: documenta la función. Puedes consultarla con `help()`:

```python
help(area_rectangulo)
```

> 💡 `print` **muestra** un valor; `return` lo **devuelve** para que puedas guardarlo y seguir usándolo. Una función sin `return` devuelve `None`.

```python
def saluda_print(nombre):
    print(f"Hola, {nombre}")

def saluda_return(nombre):
    return f"Hola, {nombre}"

a = saluda_print("Ana")
b = saluda_return("Ana")
print("a vale:", a)
print("b vale:", b)
```

## 4.2 Parámetros con valor por defecto y por nombre

```python
def precio_final(precio, iva=0.21, descuento=0):
    con_descuento = precio * (1 - descuento)
    return round(con_descuento * (1 + iva), 2)


print(precio_final(100))                       # usa los valores por defecto
print(precio_final(100, iva=0.10))             # IVA reducido
print(precio_final(100, descuento=0.15))       # solo cambiamos el descuento
```

Una función puede devolver **varios valores** (en realidad, una tupla):

```python
def estadisticas(valores):
    media = sum(valores) / len(valores)
    return min(valores), max(valores), media


minimo, maximo, media = estadisticas([4, 8, 15, 16, 23, 42])
print(f"min={minimo}, max={maximo}, media={media:.2f}")
```

## 4.3 Ámbito de las variables

Las variables creadas **dentro** de una función solo existen dentro de ella.

```python
def calcula():
    resultado_interno = 42
    return resultado_interno

calcula()
# print(resultado_interno)   # ← descomenta: NameError, no existe fuera
```

## 4.4 Errores y excepciones

Cuando algo va mal, Python lanza una **excepción** y detiene el programa. Con `try` / `except` podemos capturarla y reaccionar.

```python
def a_numero(texto):
    try:
        return float(texto)
    except ValueError:
        print(f"'{texto}' no es un número válido")
        return None


print(a_numero("3.5"))
print(a_numero("tres"))
```

Errores frecuentes que conviene reconocer:

| Excepción | Cuándo ocurre |
|---|---|
| `NameError` | Usas una variable que no existe (¿la celda no se ejecutó?) |
| `TypeError` | Operación con tipos incompatibles (`"a" + 1`) |
| `ValueError` | El tipo es correcto pero el valor no (`int("hola")`) |
| `IndexError` | Índice fuera de rango en una lista |
| `KeyError` | Clave inexistente en un diccionario |
| `ZeroDivisionError` | División entre cero |
| `IndentationError` | Sangría incorrecta |

```python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return float("inf")

dividir(10, 0)
```

## 4.5 Módulos

Python trae una enorme **biblioteca estándar**. Se usa con `import`:

```python
import math
import random
from datetime import date

print(math.sqrt(16), math.pi)
print(random.choice(["cara", "cruz"]))
print("Hoy es", date.today().strftime("%d/%m/%Y"))
```

Y en este entorno tienes instaladas librerías externas muy populares, como **pandas** (tablas de datos) o **matplotlib** (gráficos). Un pequeño adelanto:

```python
import pandas as pd

datos = pd.DataFrame({
    "alumno": ["Ana", "Luis", "Marta", "Pedro", "Eva"],
    "nota": [8.5, 4.0, 6.2, 3.5, 9.1],
})
datos["aprobado"] = datos["nota"] >= 5
datos
```

```python
import matplotlib.pyplot as plt

plt.style.use("momentum.mplstyle")  # colores y estilo del curso

colores = ["#293B8F" if ok else "#E5484D" for ok in datos["aprobado"]]  # azul: aprobado, rojo: suspenso
plt.bar(datos["alumno"], datos["nota"], color=colores)
plt.axhline(5, color="#7A8CA8", linestyle="--")
plt.title("Notas del grupo")
plt.ylabel("Nota")
plt.show()
```
