## 3.1 Condicionales

````python
nota = 7.5

if nota >= 9:
    calificacion = "Sobresaliente"
elif nota >= 7:
    calificacion = "Notable"
elif nota >= 5:
    calificacion = "Aprobado"
else:
    calificacion = "Suspenso"

print(f"Con un {nota} tienes un {calificacion}")
````

Cambia el valor de `nota` y vuelve a ejecutar la celda para ver las distintas ramas.

Para casos sencillos existe la **expresión condicional** en una línea:

````python
edad = 16
mensaje = "puede votar" if edad >= 18 else "no puede votar"
mensaje
````

> 💡 **Valores "falsos":** `0`, `0.0`, `""`, `[]`, `{}` y `None` se consideran `False` en un `if`. Por eso es habitual escribir `if lista:` para comprobar que no está vacía.

````python
carrito = []
if carrito:
    print("Procesando pedido…")
else:
    print("El carrito está vacío")
````

## 3.2 Bucle `for`

Recorre los elementos de cualquier colección, uno a uno:

````python
colores = ["rojo", "verde", "azul"]
for color in colores:
    print("Me gusta el", color)
````

`range()` genera secuencias de números — muy útil para repetir algo N veces:

````python
for i in range(5):          # 0, 1, 2, 3, 4
    print(i, end=" ")
print()

for i in range(2, 11, 2):   # desde 2 hasta 10, de 2 en 2
    print(i, end=" ")
````

`enumerate()` te da a la vez la posición y el elemento:

````python
ranking = ["Marta", "Luis", "Ana"]
for posicion, nombre in enumerate(ranking, start=1):
    print(f"{posicion}º {nombre}")
````

Un patrón muy común: **acumular** un resultado mientras recorres.

````python
precios = {"pan": 1.2, "leche": 0.95, "huevos": 2.4, "café": 4.5}

total = 0
for producto, precio in precios.items():
    if precio > 2:
        print(f"{producto} es caro ({precio} €)")
    total += precio

print(f"Total: {total:.2f} €")
````

## 3.3 `break` y `continue`

- `break` sale del bucle inmediatamente.
- `continue` salta a la siguiente vuelta.

````python
numeros = [4, 7, -1, 12, 0, 9]

for n in numeros:
    if n < 0:
        continue            # ignora los negativos
    if n == 0:
        print("Encontrado un 0, paro")
        break
    print("Procesando", n)
````

## 3.4 Bucle `while`

Repite **mientras** se cumpla una condición. Útil cuando no sabes de antemano cuántas vueltas darás.

> ⚠️ Asegúrate de que la condición acabe siendo falsa, o tendrás un bucle infinito (se para con el botón ⏹ *Interrupt*).

````python
ahorro = 0
mes = 0
while ahorro < 1000:
    ahorro += 150
    mes += 1

print(f"Llegas a 1000 € en {mes} meses (tendrás {ahorro} €)")
````

## 3.5 Comprensiones de listas

Una forma compacta de crear una lista a partir de otra:

```python
nueva = [expresión for elemento in colección if condición]
```

````python
numeros = range(1, 11)

cuadrados = [n ** 2 for n in numeros]
pares = [n for n in numeros if n % 2 == 0]

print(cuadrados)
print(pares)
````

Es equivalente a este bucle, pero en una sola línea:

````python
cuadrados = []
for n in numeros:
    cuadrados.append(n ** 2)
cuadrados
````

También existen para diccionarios:

````python
nombres = ["ana", "luis", "marta"]
longitudes = {nombre: len(nombre) for nombre in nombres}
longitudes
````
