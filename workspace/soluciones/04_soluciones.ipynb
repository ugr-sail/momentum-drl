**Ejercicio 4.1**

```python
def es_par(n):
    """Devuelve True si n es par."""
    return n % 2 == 0


for numero in [4, 7, 0, -2]:
    print(numero, es_par(numero))
```

**Ejercicio 4.2**

```python
def imc(peso, altura):
    """Índice de masa corporal redondeado a 1 decimal."""
    return round(peso / altura ** 2, 1)


def clasifica_imc(valor):
    if valor < 18.5:
        return "Bajo peso"
    elif valor < 25:
        return "Normal"
    else:
        return "Sobrepeso"


mi_imc = imc(70, 1.75)
print(mi_imc, clasifica_imc(mi_imc))
```

**Ejercicio 4.3 · Reto final**

```python
def resumen_texto(texto):
    """Cuenta palabras, palabras distintas y devuelve la más larga."""
    palabras = texto.split()
    return {
        "palabras": len(palabras),
        "distintas": len({p.lower() for p in palabras}),
        "mas_larga": max(palabras, key=len),
    }


frase = "El que lee mucho y anda mucho ve mucho y sabe mucho"
resumen_texto(frase)
```
