## 1.1 Variables

Una variable es un **nombre que apunta a un valor**. Se crea con `=` y no hace falta declarar el tipo: Python lo deduce.

Convenciones para nombres: minúsculas y guiones bajos (`precio_total`), sin empezar por número, sin tildes ni espacios.

```python
nombre = "Ana"
edad = 34
altura = 1.68
es_estudiante = False

print(nombre, edad, altura, es_estudiante)
```

En un notebook, la **última expresión** de una celda se muestra automáticamente, sin necesidad de `print()`:

```python
edad
```

Una variable puede cambiar de valor (y hasta de tipo) en cualquier momento:

```python
edad = edad + 1
edad
```

## 1.2 Tipos básicos

| Tipo | Ejemplo | Para qué |
|---|---|---|
| `int` | `42` | Números enteros |
| `float` | `3.14` | Números con decimales |
| `str` | `"hola"` | Texto |
| `bool` | `True` / `False` | Verdadero o falso |
| `NoneType` | `None` | "Sin valor" |

La función `type()` nos dice el tipo de cualquier valor:

```python
print(type(edad))
print(type(altura))
print(type(nombre))
print(type(es_estudiante))
print(type(None))
```

## 1.3 Operadores aritméticos

```python
a = 17
b = 5

print("suma        ", a + b)
print("resta       ", a - b)
print("producto    ", a * b)
print("división    ", a / b)    # siempre devuelve float
print("div. entera ", a // b)   # descarta los decimales
print("resto       ", a % b)    # módulo
print("potencia    ", a ** 2)
```

> 💡 **Cuidado con los decimales.** Los `float` se guardan en binario y a veces no son exactos. Para mostrarlos, redondea:

```python
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))
```

Forma abreviada para actualizar una variable: `+=`, `-=`, `*=`, `/=`…

```python
contador = 0
contador += 1
contador += 1
contador
```

## 1.4 Comparaciones y lógica

Las comparaciones devuelven un `bool`: `==`, `!=`, `<`, `>`, `<=`, `>=`.
Se combinan con `and`, `or` y `not`.

```python
nota = 7.5

print(nota >= 5)                  # ¿aprobado?
print(nota == 10)                 # ¿matrícula?
print(nota >= 7 and nota < 9)     # ¿notable?
print(not es_estudiante)
```

## 1.5 Cadenas de texto (`str`)

Se escriben con comillas simples o dobles. Algunas operaciones útiles:

```python
saludo = "Hola"
destino = "Granada"

frase = saludo + ", " + destino   # concatenar
print(frase)
print(len(frase))                  # longitud
print(frase.upper())               # mayúsculas
print(frase.lower())               # minúsculas
print(frase.replace("Hola", "Adiós"))
print("Gra" in frase)              # ¿contiene?
print("-" * 20)                    # repetir
```

Cada carácter tiene una posición (**índice**), empezando en `0`. Los índices negativos cuentan desde el final.

```python
palabra = "Python"
print(palabra[0])     # primera letra
print(palabra[-1])    # última letra
print(palabra[0:3])   # de la posición 0 a la 2 → "Pyt"
```

## 1.6 f-strings: la forma cómoda de formatear

Pon una `f` delante de las comillas y escribe expresiones entre `{llaves}`:

```python
producto = "café"
precio = 1.4
unidades = 3

print(f"{unidades} x {producto} = {precio * unidades} €")
print(f"Total con 2 decimales: {precio * unidades:.2f} €")
```

## 1.7 Conversión de tipos

Muchas veces el dato llega como texto (de un fichero, de un formulario…) y hay que convertirlo:

```python
texto = "42"
numero = int(texto)
print(numero + 8)

print(float("3.5") * 2)
print(str(100) + " puntos")
print(bool(0), bool(1), bool(""), bool("hola"))
```

> ⚠️ Si mezclas tipos sin convertir, Python da error. Descomenta la línea y ejecútala para verlo:

```python
# "Tengo " + 34 + " años"
```
