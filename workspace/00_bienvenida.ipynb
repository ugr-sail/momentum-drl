## Cómo usar un notebook

- Un notebook está formado por **celdas**: de texto (como esta) o de código.
- Ejecuta una celda con **Shift + Enter**. El resultado aparece justo debajo.
- Las celdas comparten memoria: una variable creada en una celda existe en las siguientes **si ya las has ejecutado**.
- Si algo se lía, usa **Restart** (arriba) y vuelve a ejecutar desde el principio.
- Si te pide elegir kernel, selecciona **Python (curso)**.
