## 2.1 Listas

La colección más usada: una secuencia ordenada y modificable.

```python
frutas = ["manzana", "pera", "plátano"]
print(frutas)
print(len(frutas))
print(frutas[0], frutas[-1])
```

Añadir, quitar y modificar elementos:

```python
frutas.append("kiwi")          # añade al final
frutas.insert(1, "fresa")      # añade en la posición 1
frutas[0] = "manzana verde"    # modifica
frutas.remove("pera")          # quita por valor
ultima = frutas.pop()          # quita y devuelve el último

print(frutas)
print("He sacado:", ultima)
```

**Slicing** (rebanadas): `lista[inicio:fin:paso]`. El `fin` **no** se incluye.

```python
numeros = [10, 20, 30, 40, 50, 60]

print(numeros[1:4])   # [20, 30, 40]
print(numeros[:3])    # los tres primeros
print(numeros[-2:])   # los dos últimos
print(numeros[::2])   # uno sí, uno no
print(numeros[::-1])  # al revés
```

Funciones y métodos útiles con listas de números:

```python
notas = [6.5, 8.0, 4.5, 9.2, 7.0]

print("máxima:", max(notas))
print("mínima:", min(notas))
print("media: ", round(sum(notas) / len(notas), 2))
print("ordenadas:", sorted(notas))
print("¿hay un 9.2?", 9.2 in notas)
```

> 💡 `sorted(lista)` devuelve una **copia** ordenada; `lista.sort()` ordena **la propia lista** y no devuelve nada.

## 2.2 Tuplas

Como una lista, pero **no se puede modificar**. Útiles para datos que van juntos y no cambian (coordenadas, fechas, registros).

```python
punto = (37.18, -3.60)    # latitud, longitud de Granada
print(punto[0])

# Desempaquetado: asignar cada elemento a una variable
lat, lon = punto
print(f"Latitud {lat}, longitud {lon}")
```

## 2.3 Diccionarios

Guardan pares **clave → valor**. Se accede por la clave, no por la posición. Ideales para describir "cosas" con atributos.

```python
alumno = {
    "nombre": "Luis",
    "edad": 22,
    "asignaturas": ["Mates", "Física"],
}

print(alumno["nombre"])
print(alumno["asignaturas"][0])
```

```python
alumno["edad"] = 23               # modificar
alumno["email"] = "luis@uni.es"   # añadir clave nueva
del alumno["asignaturas"]         # borrar

print(alumno)
print("¿Tiene email?", "email" in alumno)
```

Si la clave no existe, `alumno["x"]` da error. Con `.get()` puedes dar un valor por defecto:

```python
print(alumno.get("telefono", "sin teléfono"))
```

Recorrer claves, valores o ambos (lo veremos a fondo en el siguiente notebook):

```python
print(list(alumno.keys()))
print(list(alumno.values()))

for clave, valor in alumno.items():
    print(f"{clave:>8}: {valor}")
```

## 2.4 Conjuntos (`set`)

Colección **sin orden y sin duplicados**. Perfectos para eliminar repetidos o comparar grupos.

```python
visitas = ["ana", "luis", "ana", "marta", "luis", "ana"]
unicos = set(visitas)
print(unicos)
print(len(unicos), "visitantes distintos")
```

```python
python = {"ana", "luis", "marta"}
excel = {"luis", "pedro"}

print("ambos cursos:    ", python & excel)   # intersección
print("alguno:          ", python | excel)   # unión
print("solo Python:     ", python - excel)   # diferencia
```
