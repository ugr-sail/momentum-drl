**Ejercicio 2.1**

```python
temperaturas = [18, 21, 25, 19, 30, 27, 22]

print("máxima:", max(temperaturas))
print("mínima:", min(temperaturas))
print(f"media: {sum(temperaturas) / len(temperaturas):.1f}")
print("fin de semana:", temperaturas[-2:])
```
